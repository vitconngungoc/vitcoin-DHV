### 1. Hàm `groupby()`
* **Dùng khi nào:** Khi cần chia dữ liệu thành các nhóm dựa trên một số tiêu chí (ví dụ: nhóm theo đội bóng, nhóm theo khu vực), sau đó áp dụng một phép tính (tổng, trung bình, đếm...) lên từng nhóm đó.
* **Cú pháp cơ bản:** `df.groupby('Cột_để_nhóm')['Cột_để_tính_toán'].phép_tính()`

```python
import pandas as pd

# Ví dụ: Thống kê số bàn thắng của các cầu thủ theo từng đội bóng
data_bongda = {
    'CauThu': ['Rashford', 'Salah', 'Garnacho', 'Nunez', 'Bruno'],
    'DoiBong': ['MU', 'Liverpool', 'MU', 'Liverpool', 'MU'],
    'BanThang': [1, 2, 1, 0, 0]
}
df_bongda = pd.DataFrame(data_bongda)

# Tính tổng bàn thắng của mỗi đội
tong_ban_thang = df_bongda.groupby('DoiBong')['BanThang'].sum().reset_index()
tong_ban_thang
```

### 2. Hàm `merge()`
* **Dùng khi nào:** Khi cần nối/ghép hai bảng dữ liệu lại với nhau dựa trên một cột chung (hoạt động tương tự như hàm VLOOKUP trong Excel hoặc JOIN trong SQL).
* **Cú pháp cơ bản:** `pd.merge(bang_trai, bang_phai, on='Cột_chung', how='kiểu_nối')`

```python
# Ví dụ: Nối bảng thông tin sinh viên và bảng điểm thi dựa trên Mã sinh viên
df_sinhvien = pd.DataFrame({
    'MaSV': ['SV01', 'SV02', 'SV03'],
    'Ten': ['Tùng', 'Hải', 'Minh']
})

df_diemthi = pd.DataFrame({
    'MaSV': ['SV02', 'SV01', 'SV03'],
    'DiemJava': [8.5, 9.0, 7.5]
})

# Nối 2 bảng lại để biết sinh viên nào được mấy điểm
df_hoanchinh = pd.merge(df_sinhvien, df_diemthi, on='MaSV', how='inner')
df_hoanchinh
```

### 3. Hàm `pivot_table()`
* **Dùng khi nào:** Khi cần xoay chiều dữ liệu (reshape) để tạo ra một bảng tổng hợp đa chiều, giúp nhìn nhận dữ liệu dưới dạng ma trận (hàng và cột).
* **Cú pháp cơ bản:** `pd.pivot_table(df, values='Cột_giá_trị', index='Cột_làm_hàng', columns='Cột_làm_cột', aggfunc='phép_tính')`

```python
# Ví dụ: Thống kê số lượng món ăn khô bán được trong tựa game quản lý nhà hàng
df_nhahang = pd.DataFrame({
    'Ngay': ['Thứ 2', 'Thứ 2', 'Thứ 3', 'Thứ 3'],
    'MonAn': ['Cơm tấm', 'Gỏi cuốn', 'Cơm tấm', 'Gỏi cuốn'],
    'SoLuongBan': [50, 30, 60, 45]
})

# Tạo Pivot Table: Hàng là Ngày, Cột là Món Ăn, Giá trị là Tổng số lượng bán
bang_tong_hop = pd.pivot_table(df_nhahang, values='SoLuongBan', index='Ngay', columns='MonAn', aggfunc='sum')
bang_tong_hop
```

### 4. Hàm `apply()`
* **Dùng khi nào:** Khi cần áp dụng một hàm tự định nghĩa (hoặc hàm lambda) lên từng giá trị của một cột, hoặc từng hàng của bảng dữ liệu.
* **Cú pháp cơ bản:** `df['Cột_mới'] = df['Cột_cũ'].apply(tên_hàm)`

```python
# Ví dụ: Áp dụng mã giảm giá 10% cho danh sách các món ăn
df_gia = pd.DataFrame({
    'MonAn': ['Cơm tấm sườn', 'Gỏi cuốn tôm thịt', 'Cơm xá xíu'],
    'GiaGoc': [40000, 15000, 45000]
})

# Viết hàm tính giá sau khi giảm 10%
def giam_gia_10_phan_tram(gia):
    return gia * 0.9

# Dùng apply để tính toán hàng loạt
df_gia['GiaKhuyenMai'] = df_gia['GiaGoc'].apply(giam_gia_10_phan_tram)
df_gia
```

### 5. Hàm `melt()`
* **Dùng khi nào:** Ngược lại với Pivot. Dùng để "làm phẳng" dữ liệu, chuyển dữ liệu từ dạng ngang (wide format - nhiều cột) sang dạng dọc (long format - ít cột hơn nhưng nhiều hàng hơn). Rất hữu ích khi cần vẽ biểu đồ.
* **Cú pháp cơ bản:** `pd.melt(df, id_vars=['Cột_giữ_nguyên'], value_vars=['Các_cột_muốn_chuyển_thành_hàng'], var_name='Tên_cột_chứa_tiêu_đề', value_name='Tên_cột_chứa_giá_trị')`

```python
# Ví dụ: Chuyển đổi bảng báo cáo doanh thu của 3 vùng: Bắc, Trung, Nam từ dạng ngang sang dọc
df_doanhthu = pd.DataFrame({
    'Quy': ['Quý 1', 'Quý 2'],
    'MienBac': [150, 200],
    'MienTrung': [100, 120],
    'MienNam': [300, 350]
})

# Melt để gộp 3 cột miền thành 1 cột "KhuVuc"
df_melt = pd.melt(df_doanhthu, 
                  id_vars=['Quy'], 
                  value_vars=['MienBac', 'MienTrung', 'MienNam'],
                  var_name='KhuVuc', 
                  value_name='DoanhThu')
df_melt
```

### 6. Hàm `sort_values()`
* **Dùng khi nào:** Khi cần sắp xếp bảng dữ liệu tăng dần hoặc giảm dần theo một hoặc nhiều cột (ví dụ: xếp hạng điểm số, xếp hạng doanh thu).
* **Cú pháp cơ bản:** `df.sort_values(by='Tên_cột', ascending=True_hoặc_False)`

```python

# Ví dụ: Sắp xếp danh sách cầu thủ theo số bàn thắng giảm dần
df_sort = pd.DataFrame({
    'CauThu': ['Rashford', 'Hojlund', 'Garnacho'],
    'BanThang': [5, 12, 7]
})

# Sắp xếp giảm dần (ascending=False)
bang_xep_hang = df_sort.sort_values(by='BanThang', ascending=False)
bang_xep_hang
```

### 7. Hàm `isna()`
* **Dùng khi nào:** Khi cần kiểm tra xem trong bảng dữ liệu có ô nào bị trống (thiếu dữ liệu / NaN) hay không. Thường dùng chung với `.sum()` để đếm số lượng ô trống.
* **Cú pháp cơ bản:** `df.isna().sum()`

```python
import numpy as np

# Ví dụ: Kiểm tra xem có sinh viên nào bị thiếu điểm không
df_diem = pd.DataFrame({
    'Ten': ['Tùng', 'Hải', 'Minh'],
    'Diem': [9.0, np.nan, 7.5] # Hải bị thiếu điểm
})

# Đếm số lượng ô bị trống trong mỗi cột
kiem_tra_thieu_du_lieu = df_diem.isna().sum()
kiem_tra_thieu_du_lieu
```

### 8. Hàm `drop_duplicates()`
* **Dùng khi nào:** Khi dữ liệu bị lỗi trùng lặp (ví dụ: một người nhập đơn hàng 2 lần) và bạn muốn xóa các dòng giống hệt nhau, chỉ giữ lại 1 dòng duy nhất.
* **Cú pháp cơ bản:** `df.drop_duplicates()`

```python
# Ví dụ: Xóa các đơn hàng bị ghi trùng lặp
df_donhang = pd.DataFrame({
    'MaDon': ['DH01', 'DH02', 'DH01'], # DH01 bị lặp lại
    'MonAn': ['Cơm tấm', 'Phở', 'Cơm tấm']
})

# Xóa dòng trùng
df_sach = df_donhang.drop_duplicates()
df_sach
```

## NumPy Cheat Sheet

### 1. Hàm `np.where()`
* **Dùng khi nào:** Hoạt động giống hệt hàm IF trong Excel. Dùng để kiểm tra một điều kiện, nếu đúng trả về giá trị A, nếu sai trả về giá trị B.
* **Cú pháp cơ bản:** `np.where(điều_kiện, giá_trị_nếu_đúng, giá_trị_nếu_sai)`

```python
# Ví dụ: Xét kết quả Đậu/Rớt dựa trên điểm số
diem_so = np.array([4.5, 8.0, 3.0, 9.5])

# Nếu điểm >= 5.0 thì "Đậu", ngược lại "Rớt"
ket_qua = np.where(diem_so >= 5.0, 'Đậu', 'Rớt')
ket_qua
```

### 2. Hàm `np.arange()`
* **Dùng khi nào:** Khi cần tạo nhanh một mảng các số theo một quy luật nhất định (có số bắt đầu, số kết thúc và bước nhảy).
* **Cú pháp cơ bản:** `np.arange(start, stop, step)`

```python
# Ví dụ: Tạo một mảng gồm các số chẵn từ 0 đến 10 (không bao gồm 10)
so_chan = np.arange(0, 10, 2)
so_chan
```
