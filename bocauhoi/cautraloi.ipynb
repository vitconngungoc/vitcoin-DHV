+ SV-A dùng dropna(): Loại bỏ 177 hàng (20%). Histogram chỉ vẽ trên 714 hàng còn lại. Nếu những người thiếu tuổi không phân phối ngẫu nhiên, histogram sẽ thiếu một vùng tuổi nhất định, phân phối trông "cắt cụt" ở một đầu.
+ SV-B dùng fillna(mean): Thêm 177 giá trị đều bằng đúng 1 con số. Histogram sẽ xuất hiện một cột đột biến cao bất thường tại giá trị mean, trong khi các bin xung quanh bình thường. Đây là dấu hiệu nhận biết rõ ràng của fillna(mean).

+ Ai tạo ra bias lớn hơn:
SV-B tạo ra bias lớn hơn trong trường hợp này. Lý do: fillna(mean) làm méo hình dạng phân phối — tạo ra một đỉnh nhân tạo tại mean, kéo độ lệch chuẩn xuống thấp hơn thực tế, và làm mọi phân tích phân phối (histogram, box plot, skewness) trở nên không tin cậy. SV-A mất dữ liệu nhưng những gì còn lại là thật; SV-B giữ đủ số hàng nhưng đưa vào dữ liệu giả.

+ Trường hợp có thể chấp nhận dropna():
