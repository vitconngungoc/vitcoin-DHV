## 6. Summary Statistics and Outlier Detection
Thống kê mô tả cho các cột số và kiểm tra các giá trị ngoại lai hoặc bất thường.

```python
# Thống kê mô tả cho các cột số
summary = df.describe()
display(summary)

# Kiểm tra các giá trị ngoại lai bằng IQR cho các cột số
for col in df.select_dtypes(include=[np.number]).columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    outliers = df[(df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)]
    print(f"Cột {col} có {len(outliers)} giá trị ngoại lai (theo IQR)")
```
